```python
import pandas as pd
df = pd.read_csv(r"C:\Users\M S\Downloads\teaching_prices_messy.csv")
df_clean = df.copy()
```

```python
print("BEFORE CLEANING")
print(f"Rows: {df.shape[0]}")
print(f"Missing prices: {df['price'].isna().sum()}")
print(f"Unique items: {df['item'].unique().tolist()}")
print(f"Duplicate rows: {df.duplicated().sum()}")
print(f"Price dtype: {df['price'].dtype}")
print()
print(df.describe())
```

```python
df_clean = df_clean.drop_duplicates(keep='first')
print(f"Rows after dropping duplicates: {df_clean.shape[0]}")
```

```python
mask = df_clean['item'] == 'oil b'
df_clean.loc[mask, 'item'] = 'Staple A'
print("Fixed 'oil b' → 'Staple A'")
```

```python
df_clean.loc[df_clean['price'] == '9999', 'price'] = None
print("Replaced 9999 with NaN")
```

```python
df_clean.loc[df_clean['price'].str.contains(',', na=False), 'price'] = None
print("Replaced '1,250' with NaN")
```

```python
df_clean['price'] = pd.to_numeric(df_clean['price'], errors='coerce')
print(f"Price dtype now: {df_clean['price'].dtype}")
```

```python
print("AFTER CLEANING")
print(f"Rows: {df_clean.shape[0]}")
print(f"Missing prices: {df_clean['price'].isna().sum()}")
print(f"Unique items: {df_clean['item'].unique().tolist()}")
print(f"Duplicate rows: {df_clean.duplicated().sum()}")
print(f"Price dtype: {df_clean['price'].dtype}")
print()
print(df_clean.describe())
```

```python
print("Rows per date (expect 8 each):")
print(df_clean['date'].value_counts().sort_index())
print()
print("Rows per item (expect 24 each):")
print(df_clean['item'].value_counts())
print()
print("Missing price rows:")
print(df_clean[df_clean['price'].isna()][['date','item','city','price']])
```

```python
comparison = pd.DataFrame({
    'Metric': ['Total rows', 'Duplicate rows', 'Missing prices', 'Unique items', 'Price dtype'],
    'Before': [97, 1, 1, 5, 'string'],
    'After': [96, 0, 3, 4, 'float64']
})
print(comparison.to_string(index=False))
```
